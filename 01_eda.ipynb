# Exploratory Data Analysis — Logistics Orders

Explores order volume trends, delivery time distributions, and warehouse-level patterns ahead of demand forecasting and route optimization.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv("../data/processed/orders_clean.csv", parse_dates=["order_date", "delivery_date"])
df.head()
```

## Order volume over time

```python
daily_orders = df.groupby(df["order_date"].dt.date).size()
daily_orders.plot(figsize=(10, 4), title="Daily Order Volume")
plt.ylabel("Orders")
plt.show()
```

## Delivery time distribution

```python
sns.histplot(df["delivery_hours"], bins=30, kde=True)
plt.title("Distribution of Delivery Time (hours)")
plt.xlabel("Hours")
plt.show()
```

## On-Time Delivery Rate by warehouse

```python
if "on_time" in df.columns:
    otdr_by_warehouse = df.groupby("warehouse_id")["on_time"].mean().sort_values() * 100
    otdr_by_warehouse.plot(kind="barh", figsize=(8, 4), title="On-Time Delivery Rate by Warehouse (%)")
    plt.xlabel("OTDR (%)")
    plt.show()
```

## Delivery locations (geographic density)

```python
plt.figure(figsize=(6, 6))
plt.scatter(df["delivery_lon"], df["delivery_lat"], s=5, alpha=0.4)
plt.title("Delivery Locations")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.show()
```
